```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

### Matplotlib Basics

```python
x = np.linspace(0,5,11)
```

```python
x
```

```python
y = x**2
```

```python
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
```

### SubPlot in Matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

### Object Oriented Way

```python
fig = plt.figure()
fig.add_axes([0.1,0.1,0.8,0.8]) #left shift,bottom shift,width,height
```

```python
fig = plt.figure(figsize = (20,8),dpi=100) #dpi= depth per inch
axis_1 = fig.add_axes([.1,.1,.4,.8])
axis_1.plot(x,y)
axis_2 = fig.add_axes([.3,.4,.3,.3])
axis_2.plot(x,y)
```

### Types Of Plots

```python
plt.scatter(x,y)
```

```python
plt.hist(x) 
```

```python
plt.boxplot(x)
```

### Saving Plots

```python
fig = plt.figure()
ax =fig.add_axes([.1,.1,.8,.8])
ax.plot(x,y)
plt.savefig("basic plot.png")
```

```python
plt.plot(x,y,color = "pink")
```
